```python
import networkx as nx
import pandas as pd
import numpy as np
from pathlib import Path

# Load nodes and edges from CSV
nodes_df = pd.read_csv("nodes.csv")
edges_df = pd.read_csv("edges.csv")

# Create a NetworkX graph from the CSV data
G = nx.DiGraph()

# Add nodes with attributes
for _, row in nodes_df.iterrows():
    G.add_node(
        int(row["id"]),
        name=row["name"],
        zone=row["zone"],
        base_kv=row["base_kv"],
        lat=row["lat"],
        lng=row["lng"],
        mw_load=row["mw_load"],
        gen_mw=row["gen_mw"],
        gen_fuels=row["gen_fuels"]
    )

# Add edges with attributes
for _, row in edges_df.iterrows():
    G.add_edge(
        int(row["from_bus"]),
        int(row["to_bus"]),
        uid=row["uid"],
        voltage_kv=row["voltage_kv"],
        length_km=row["length_km"],
        capacity_mva=row["capacity_mva"],
        reactance=row["reactance"]
    )

# Convert to undirected graph for analysis (since power flows both ways)
G = G.to_undirected()

# Extract the giant connected component
largest_cc = max(nx.connected_components(G), key=len)
G_gcc = G.subgraph(largest_cc).copy()

print(f"Network loaded: {G_gcc.number_of_nodes()} nodes, {G_gcc.number_of_edges()} edges")
print(f"Removed {G.number_of_nodes() - G_gcc.number_of_nodes()} isolated nodes")
```
