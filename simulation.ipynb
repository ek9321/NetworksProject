```python
import networkx as nx
import pandas as pd
import numpy as np
from pathlib import Path

# Load the ERCOT network graph
graphml_path = Path("ercot_network.graphml")
G = nx.read_graphml(graphml_path)

# Convert node IDs back to integers
G = nx.relabel_nodes(G, {node: int(node) for node in G.nodes()})

# Extract the giant connected component
largest_cc = max(nx.connected_components(G), key=len)
G_gcc = G.subgraph(largest_cc).copy()

print(f"Network loaded: {G_gcc.number_of_nodes()} nodes, {G_gcc.number_of_edges()} edges")
print(f"Removed {G.number_of_nodes() - G_gcc.number_of_nodes()} isolated nodes")
```
